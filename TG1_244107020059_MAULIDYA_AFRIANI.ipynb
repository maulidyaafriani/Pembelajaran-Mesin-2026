## Bagian 2 - No. 1: Contoh Pelanggaran Etika dan Hukum dalam Penggunaan AI

### Contoh 1: Pengumpulan data wajah tanpa persetujuan (Clearview AI)
Clearview AI mengumpulkan miliaran foto wajah dari internet dan media sosial
tanpa izin pemiliknya untuk membangun basis data pengenalan wajah yang dijual
kepada penegak hukum. Otoritas perlindungan data di beberapa negara menyatakan
praktik ini melanggar hukum privasi. Regulator Italia (Garante) menjatuhkan
denda 20 juta euro pada 2022, dan regulator Belanda (Dutch DPA) menjatuhkan
denda 30,5 juta euro pada 2024.

- Pelanggaran etika: melanggar privasi dan otonomi individu karena tidak ada
  persetujuan (*informed consent*).
- Pelanggaran hukum: bertentangan dengan GDPR di Uni Eropa. Di Indonesia,
  praktik serupa berpotensi melanggar UU No. 27 Tahun 2022 tentang
  Pelindungan Data Pribadi (UU PDP).

### Contoh 2: Penipuan dengan deepfake (kasus Arup, Hong Kong)
Pada 2024, seorang karyawan keuangan di kantor Arup Hong Kong mentransfer
sekitar 25 juta dolar AS setelah mengikuti video call yang ternyata dipalsukan
dengan deepfake, di mana sosok CFO dan rekan kerja ditiru menggunakan AI.

- Pelanggaran etika: manipulasi dan penipuan dengan meniru identitas orang lain.
- Pelanggaran hukum: penipuan dan pemalsuan identitas. Di Indonesia dapat
  dijerat UU ITE (manipulasi informasi/dokumen elektronik).

### Contoh 3: Menyerahkan output AI tanpa verifikasi (Mata v. Avianca)
Pada 2023, pengacara di New York mengajukan dokumen pengadilan yang berisi
kutipan kasus fiktif hasil karangan ChatGPT. Hakim menjatuhkan sanksi kepada
para pengacara tersebut.

- Pelanggaran etika: tidak ada verifikasi dan pertanggungjawaban atas keluaran AI
  (*hallucination*).
- Pelanggaran hukum/profesi: melanggar kewajiban profesional kepada pengadilan.
